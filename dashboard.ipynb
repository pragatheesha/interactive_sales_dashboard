```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# Load data
df = pd.read_csv("/content/sales_data.csv")
df
```

```python
# Seaborn theme
sns.set_theme(style="whitegrid", palette="Set2")
```

```python
import os

# Create the 'visualizations' directory if it doesn't exist
os.makedirs('visualizations', exist_ok=True)

# 1. BOX PLOT – Price Distribution

plt.figure()
sns.boxplot(x="Product", y="Price", data=df)
plt.title("Price Distribution by Product")
plt.savefig("visualizations/boxplot_price.png")
plt.show()
```

```python
# 2. BAR CHART – Sales by Product
# -----------------------------
plt.figure()
sns.barplot(x="Product", y="Total_Sales", data=df, estimator=sum)
plt.title("Total Sales by Product")
plt.savefig("visualizations/bar_sales.png")
plt.show()
```

```python
# 3. HEATMAP – Correlation
# -----------------------------
plt.figure()
corr = df[["Quantity", "Price", "Total_Sales"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.savefig("visualizations/heatmap_corr.png")
plt.show()
```

```python
# 4. VIOLIN PLOT – Sales Distribution
# -----------------------------
plt.figure()
sns.violinplot(x="Product", y="Total_Sales", data=df)
plt.title("Sales Distribution by Product")
plt.show()
```

```python
# 5. INTERACTIVE PLOT (PLOTLY)
# -----------------------------
fig = px.bar(
    df,
    x="Product",
    y="Total_Sales",
    color="Region",
    title="Interactive Sales by Product",
    hover_data=["Quantity", "Price"]
)
fig.show()
```
