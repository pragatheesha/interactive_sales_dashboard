```python
# Import
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
# Create data

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Laptop", "Mouse"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics", "Accessories"],
    "Quantity": [2, 5, 3, 1, 1, 4],
    "Price": [500, 20, 30, 200, 500, 20],
    "Sales": [1000, 100, 90, 200, 500, 80]
}

df = pd.DataFrame(data)
df
```

```python
# Create Box Plots (Data Distribution)

sns.set(style="whitegrid")

plt.figure()
sns.boxplot(x="Product", y="Price", data=df)
plt.title("Price Distribution by Product")
plt.xlabel("Product")
plt.ylabel("Price")
plt.show()
```

```python
# Build Correlation Heatmap (Numerical Data)

plt.figure()
corr = df[["Quantity", "Price", "Sales"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Professional Charts with Custom Styles

sns.set_theme(
    style="darkgrid",
    palette="Set2",
    font_scale=1.1
)

plt.figure()
sns.barplot(x="Product", y="Sales", data=df, estimator=sum)
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
# Create Multiple Subplots (Single Figure)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

sns.boxplot(x="Category", y="Price", data=df, ax=axes[0, 0])
axes[0, 0].set_title("Price Distribution")

sns.barplot(x="Product", y="Sales", data=df, estimator=sum, ax=axes[0, 1])
axes[0, 1].set_title("Sales by Product")

sns.violinplot(x="Category", y="Sales", data=df, ax=axes[1, 0])
axes[1, 0].set_title("Sales Distribution")

sns.heatmap(df[["Quantity", "Price", "Sales"]].corr(), annot=True, ax=axes[1, 1])
axes[1, 1].set_title("Correlation")

plt.tight_layout()
plt.show()
```

```python
# Build Interactive Charts (Hover Info – Plotly)

fig = px.bar(
    df,
    x="Product",
    y="Sales",
    color="Category",
    hover_data=["Quantity", "Price"],
    title="Interactive Sales by Product"
)

fig.show()
```

```python
# Combine Charts into a Dashboard Layout

fig = px.scatter(
    df,
    x="Price",
    y="Sales",
    color="Category",
    size="Quantity",
    hover_name="Product",
    title="Interactive Sales Dashboard"
)

fig.show()
```
